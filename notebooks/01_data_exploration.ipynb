```python
df = df.drop_duplicates()
```

```python
print("Positive quantity:", (df["Quantity"] > 0).sum())
print("Negative quantity:", (df["Quantity"] < 0).sum())
print("Zero quantity:", (df["Quantity"] == 0).sum())
```

```python
purchase_df = df[df["Quantity"] > 0].copy()
```

```python
purchase_df.shape
```

```python
purchase_df["Customer ID"].isnull().sum()
```

```python
(purchase_df["Customer ID"].isnull().sum() / len(purchase_df)) * 100
```

```python
customer_df = purchase_df[purchase_df["Customer ID"].notnull()].copy()
```

```python
customer_df.shape
```

```python
customer_df["Customer ID"].nunique()
```

```python
customer_df["Revenue"] = customer_df["Quantity"] * customer_df["Price"]
```

```python
customer_df[["Quantity", "Price", "Revenue"]].head()
```

```python
customer_df["InvoiceDate"].max()
```

```python
reference_date = customer_df["InvoiceDate"].max()
```

```python
last_purchase = customer_df.groupby("Customer ID")["InvoiceDate"].max()
```

```python
recency = (reference_date - last_purchase).dt.days
```

```python
recency.head()
```

```python
frequency = customer_df.groupby("Customer ID")["Invoice"].nunique()
```

```python
frequency.head()
```

```python
monetary = customer_df.groupby("Customer ID")["Revenue"].sum()
```

```python
monetary.head()
```

```python
rfm = pd.concat([recency, frequency, monetary], axis=1)
```

```python
rfm.columns = ["Recency", "Frequency", "Monetary"]
```

```python
rfm.head()
```

```python
rfm.describe()
```

```python
import matplotlib.pyplot as plt

plt.hist(rfm["Monetary"], bins=50)
plt.xlabel("Monetary Value")
plt.ylabel("Number of Customers")
plt.title("Distribution of Monetary Value")
plt.show()
```

```python
import numpy as np

rfm['Monetary_Log'] = np.log1p(rfm['Monetary'])
```

```python
plt.figure(figsize=(8,5))

plt.hist(rfm['Monetary_Log'], bins=50)

plt.title('Distribution of Log-Transformed Monetary Value')
plt.xlabel('Log Monetary Value')
plt.ylabel('Number of Customers')

plt.show()
```
